```python
import random as random
import math as math

def coin_toss():
    r=random.random()
    mid=0.5
    if r>mid:
        return 'Tail'
    else:
        return 'Head'

head_count=0
for _ in range(10000):
    if coin_toss()=='Head':
        head_count+=1
print(f"Heads:{head_count}/10000")

def dist(a,b):
    ax,ay=a
    bx,by=b
    return math.sqrt((bx-ax)**2+(by-ay)**2)

def closer_point(a,p,b):
    pa=dist(a,p)
    pb=dist(b,p)
    if pa>pb:
        return 'B is closer'
    elif pb>pa:
        return 'A is closer'
    else:
        return 'Midway'

closer_point((1,2),(0,0),(5,5))

def find_nearest_1d(arr,num):
    d=float('inf')
    ans=[]
    if not isinstance(arr, list):
       return None
    for i in range(len(arr)):
        if(abs(arr[i]-num)<d):
            d=abs(arr[i]-num)
            ans=[arr[i]]
        elif(abs(arr[i]-num)==d):
             ans.append(arr[i])
    return ans

def find_nearest_2d(arr,target):
    d=float('inf')
    ans=[]
    if not isinstance(arr, list):
       return None
    for i,j in arr:
        if(((i-target[0])**2+(j-target[1])**2)<d):
            d=((i-target[0])**2+(j-target[1])**2)
            ans=[(i,j)]
        elif(((i-target[0])**2+(j-target[1])**2)==d):
            ans.append((i,j))
            
    return ans

def distance_nd(A,B):
    ans=0
    for i in range(len(A)):
        ans+=A[i]**2+B[i]**2
    return ans

def find_nearest_nd(points,target):
    d=float('inf')
    ans=[]

    if isinstance(target, (int, float)):
        for point in points:
            dist = (point - target) ** 2

            if dist < d:
                d = dist
                ans = [point]

            elif dist == d:
                ans.append(point)

        return ans
    
    for point in points:
        dist=0
        for i in range(len(point)):
            dist+=(point[i]-target[i])**2
        if(dist<d):
           d=dist
           ans=[point]
        elif(dist==d):
            ans.append(point)
    return ans

print(find_nearest_nd([(1, 2), (5, 6), (3, 4), (10, 10)],(4, 5)))
print(find_nearest_nd([1,2,5,7],6))
```
