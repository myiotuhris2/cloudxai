```python
def print_digit_in_one_line(num):
    if num==0:
        return
    print(f'num:{num}') 
    last_digit=num%10
    print(f'last_digit:{last_digit}')
    print_digit_in_one_line(num//10)
print_digit_in_one_line(321)


def base_8_to_base_10(num,multiplier=0,sol=0):
    if num < 0:
        return -base_8_to_base_10(-num, multiplier, sol)
    if num==0:
        return sol
    last_digit=num%10
    sol+=last_digit*8**multiplier
    return base_8_to_base_10(num//10,multiplier+1,sol)
print(base_8_to_base_10(50))

def rev_str(word,rev=''):
    if(type(word))!=str:
        return None
    if len(word)==0:
        return rev
    rev+=word[len(word)-1]
    return rev_str(word[:-1],rev)
 

def base_10_to_base_8(num,sol=''):
    if num < 0:
        return -base_10_to_base_8(-num)
    if num==0:
        return 0 if sol == '' else int(rev_str(sol, '')) #sol[::-1]
    sol+=str(num%8)
    return base_10_to_base_8(num//8,sol)

print(base_10_to_base_8(8))
```
