```python
def find_sqrt(num):
    if num < 1:
        return 'Error'
    lo,hi=1,num
    guess=(lo+hi)/2
    while True:
        if abs(guess*guess-num)<0.00001:
            return guess
        elif guess*guess>num:
            hi=guess
        else:
            lo=guess
        guess=lo+(hi-lo)/2
        # print(f"High :{hi}, Low:{lo}")
        # print(f"Guess: {guess}")
        # print(f"Square: {guess*guess}")
def hyp(b,h):
    return find_sqrt((b**2+h**2))

def find_distance_2d(x1,y1,x2,y2):
    return hyp((x2-x1),(y2-y1))

def find_distance_3d(x1,y1,z1,x2,y2,z2):
    base=hyp((x2-x1),(y2-y1))
    height=z2-z1
    return hyp(base,height)

def find_distance_3d_formula(x1,y1,z1,x2,y2,z2):
    return find_sqrt((x2-x1)**2+(y2-y1)**2+(z2-z1)**2)

def find_distance_4d_formula(x1,y1,z1,d1,x2,y2,z2,d2):
    return find_sqrt((x2-x1)**2+(y2-y1)**2+(z2-z1)**2+(d2-d1)**2)

def find_distance_4d(x1,y1,z1,d1,x2,y2,z2,d2):
    base_3d=hyp((x2-x1),(y2-y1))
    height_3d=z2-z1
    base_4d=hyp(base_3d,height_3d)
    height_4d=d2-d1
    return hyp(base_4d,height_4d)

def find_nearest_1d(arr,num):
    d=float('inf')
    n=0
    for i in range(len(arr)):
        if(abs(arr[i]-num)<d):
            d=abs(arr[i]-num)
            n=arr[i]
    return n


print(find_nearest_1d([1,5,8,10],6))

find_distance_2d(0,0,3,4)
find_distance_3d(0,0,0,1,1,1)
find_distance_4d(0,0,0,0,1,1,1,1)
find_distance_4d_formula(0,0,0,0,1,1,1,1)
```
