```python
import math
class Person:
    def __init__(self):
        self.north=0
        self.east=0

    def move(self,north,east):
        self.north=self.north+north
        self.east=self.east+east

    def distance_from_home(self):
        return math.sqrt(self.north**2+self.east**2)

person1 = Person()
person1.move(-10,0)
person1.move(0,20)
person1.move(-2,0)
person1.move(0,-3.5)

 

print(person1.distance_from_home())

```

```python
import math


class Vector2D:
    def __init__(self):
        self.x=0
        self.y=0

    def mod(self):
        return math.sqrt(self.x**2+self.y**2)

    def scale(self,times):
        self.x=self.x*times
        self.y=self.y*times

    def multiply(self,x_scale,y_scale):
        self.x=self.x*x_scale
        self.y=self.y*y_scale

class TwoPointsLine():
    def __init__(self):
        self.m=0
        self.c=0

    def fit(self,a,b):
        self.m=(b[1]-a[1])/(b[0]-a[0])
        self.c=a[1]-self.m*a[0]

    def predict(self,c):
        return self.m*c+self.c

model=TwoPointsLine()
model.fit([2,5],[3,5])
print(model.predict(3))


class Animal:
    def eat(self):
        print('Eating!')

class Dog(Animal) :
    def bark(self):
        print('Woof')
        self.eat()

dog=Dog()
dog.bark()

        
```
