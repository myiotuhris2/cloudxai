```python
def find_sqrt(num):
    if num < 1:
        return 'Error'
    lo,hi=1,max(1,float(num))
    guess=(lo+hi)/2
    while True:
        if abs(guess*guess-num)<0.00001:
            return guess
        elif guess*guess>num:
            hi=guess
        else:
            lo=guess
        guess=lo+(hi-lo)/2
        print(f"High :{hi}, Low:{lo}")
        print(f"Guess: {guess}")
        print(f"Square: {guess*guess}")
find_sqrt(10)
```
