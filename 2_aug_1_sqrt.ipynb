```python
def find_sqrt(num):
    if num < 1:
        return 'Error'
    lo,hi=1,max(1,float(num))
    guess=(lo+hi)/2
    while True:
        if abs(guess*guess-num)<0.00001:
            return guess
        elif guess*guess>num:
            hi=guess
        else:
            lo=guess
        guess=lo+(hi-lo)/2
        print(f"High :{hi}, Low:{lo}")
        print(f"Guess: {guess}")
        print(f"Square: {guess*guess}")
find_sqrt(10)

# left=0
# right=7
# def find_nroot():
#     for i in range(100):
#         if(x**x>49):
#             right=x
#         else:
#             left=x
#         print(x,x**x)


def pow(x,n,i=1,result=1):
    if(n==0):
        return 1
    if(i==abs(n)):
        if(n<0):
            return 1/(result*x)
        return result*x
    result=result*x
    return pow(x,n,i+1,result)

print(pow(10,-4))
```
