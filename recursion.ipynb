```python
def hcfcal(a,b):
    if(a==0 or b==0):
        return(max(a,b))
    if(b==1):
        return b
    if(a%b==0):
        return b
    return hcfcal(b,a%b)

print(hcfcal(max(36,0),min(36,0)))

def rec_div(dividend,divisor):
    if(divisor==0):
        return 'Arithmetic Error'
    if(dividend<divisor):
        return 0,dividend
    q,r=rec_div(dividend-divisor,divisor)
    return q+1,r

print(rec_div(7,0))
```
