```python
def find_min(arr):
    min=1e18
    for i in arr:
        if(i<min):
            min=i
    return min
print(find_min([1,2,3,4,5]))


#normalize-(arr[i]-min/max-min)
#median ignores outliers

from math import sqrt
import numpy as np
def sdev(arr):
    arr=np.array(arr)
    mean=np.mean(arr)
    sd=0
    var=0
    for i in arr:
        var+=(i-mean)**2
    var=var/arr.size
    sd=sqrt(var)
    print(var)
    print(sd)
    return sd

sdev([10,12,12,13,12,11,90])

def find_outliers(nums,threshold):
    o=[]
    sd=sdev(nums)
    mean=np.mean(np.array(nums))
    for i in nums:
        z=abs(i-mean)/sd
        if(z>abs(threshold)):
            o.append(i)
    return o
print(find_outliers([1,2,3,4,4,5,100],2))
```
