```python
def find_m_c(x1,y1,x2,y2):
    #change in y per change in x (y2-y1)/(x2-x1)
    #change in y when x1 becomes 0 y1-x1 * change in y per change in x
    m=(y2-y1)/(x2-x1)
    c=y1-m*x1
    #y=y1+(x-x1)*m
    #increase in y=(x-x1)*m
    #m->change in y per change in x
    return m,c

def predict(m,c,x):
    return m*x+c

m,c=find_m_c(1,3,2,7)
y=predict(m,c,0.5)
y
```
