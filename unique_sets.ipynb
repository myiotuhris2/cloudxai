```python
#S=1+2+...+n
#S=n+n-1+...+(n-(n-1))
#S=n*n-(1+2+...+n-1)
#S=n*n-(1+2+...+n-1+n-n)
#S=n*n-(S-n)
#2S=n^2+n
#S=n*(n+1)/2

#unique sets of r elements*orderings per r elements= total orderings of n elements taken r at a time
#unique sets of r elements*(1st place can be filled in r ways*2nd place in (r-1) ways*....*rth place can be filled in 1 way)=1st place can be filled in n ways*2nd place in (n-1) ways*....*rth place can be filled in (n-(r-1))
#unique sets of r elements*r!=n*(n-1)*...*(n-r+1)
#unique sets of r elements*r!=n*(n-1)*...*(n-r+1)*(n-r)*(n-r-1)*...*3*2*1/((n-r)*(n-r-1)*...*3*2*1)
#unique sets of r elements*r!=n!/(n-r)!
#unique sets of r elements=n!/(r!*(n-r)!)

def fact(n):
   if(n==0 or n==1):
      return 1;
   return (n*fact(n-1))

def uniquesets(n,r):
   return fact(n)/(fact(r)*fact(n-r))

us=uniquesets(4,3)
print(us)

```

```python
def combinations(arr,r,start=0,current=""):
    if(len(current)==r):
      print(current)
      return
    for i in range(start,len(arr)):
      combinations(arr,r,i+1,current+arr[i])

combinations(['A','B','C','D'], 3,0,'')
```
